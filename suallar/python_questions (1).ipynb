#### 1. Verilmiş mətndəki bütün sözləri kiçik hərfə çevirin və təkrarsız sözlərin əlifba sırası ilə siyahısını qaytarın.
###### Misal: text = "Python is great Python is easy and data is everywhere"
###### Cavab: ['and', 'data', 'easy', 'everywhere', 'great', 'is', 'python']


```python
def siyahi(text):
    b=[]
    for i in text.lower().split(' '):
        if i not in b:
            b.append(i)
    print(b)
siyahi('Python is great Python is easy and data is everywhere')    
          

```

#### 4. List comprehension istifadə edərək Selsi ilə verilmiş temperaturlardan yalnız müsbət olanları Farengeytə çevirin. Düstur: F = C × 9/5 + 32 
###### Misal: celsius = [-5, 0, 10, 20, 30, -10, 15]
###### Cavab: [50.0, 68.0, 86.0, 59.0]


```python
def selsi(celsius):
    b=[(c*9)/5+32 for c in celsius if c>=0]
    print(b)
selsi([-5,0,10,20,30,-10,15])    
```

#### 8. Rekursiyadan istifadə edərək iç-içə listləri tək düz listə çevirən `flatten` funksiyası yazın. 
###### Misal: flatten([1, [2, 3], [4, [5, 6]], [[7]], 8])
###### Cavab: [1, 2, 3, 4, 5, 6, 7, 8]

```python
def flatten(listed):
    b=[]
    for i in listed:
        if isinstance(i,list):
            b.extend(flatten(i))
        else:
            b.append(i)
    print(b)
flatten([1, [2, 3], [4, [5, 6]], [[7]], 8])               
```

#### 11. Listin içində verilmiş string-ləri uzunluqlarına görə dictionary-də qruplaşdırın.
###### Misal: strings = ["abc", "de", "fgh", "ij", "klmno", "p"]
###### Cavab: {3: ['abc', 'fgh'], 2: ['de', 'ij'], 5: ['klmno'], 1: ['p']}


```python
def siyahi(strings):
    b={}
    for i in strings:
        if i not in b:
            b[len(i)]=[]
        b[len(i)].append(i) 
    print(b)    
siyahi( ["abc", "de", "fgh", "ij", "klmno", "p"])       
```

#### 12. `Dataset` adlı class yaradın. Class ad (`name`) və ədədlər listi (`values`) qəbul etsin. Aşağıdakı metodları yazın: `size()` dəyərlərin sayını, `mean()` isə ortasını qaytarsın. 
###### Misal: ds = Dataset('temperatur', [10, 20, 30, 40, 50])     ds.size(), ds.mean()
###### Cavab: 5   30.0
###### İpucu: dəyərləri `__init__`-də `self.values`-a yazın

```python
class Dataset:
    def __init__(self,name,values):
        self.name=name
        self.values=values
    def size(self):
        print(len(self.values))
    def mean(self):
        print(self.values[len(self.values)//2])
ds = Dataset('temperatur', [10, 20, 30, 40, 50]) 
ds.size()
ds.mean()      
                  
```
